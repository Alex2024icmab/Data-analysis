```python
#Extract transfer csv files for transfer
#Extract csv files from hdf5 where VGS and IDS columns are in the same file.
import h5py  # h5py is the library used to read, write, and manipulate HDF5 files in Python
import numpy as np  # numpy: a powerful numerical computing library in Python, often used for working with arrays and matrices.

# Specify the name of your HDF5 file
filename = "Name of the file.hdf5"

# Opening and reading the HDF5 file
with h5py.File(filename, 'r') as file:  # Open the specified .hdf5 file (filename) in read-only mode ('r')
    for lv1_group_name in file.keys():  # Loop over level 1 groups
        lv1_group = file[lv1_group_name]
        for lv2_group_name in lv1_group.keys():  # Loop over level 2 groups
            lv2_group = lv1_group[lv2_group_name]
            if 'measured_I_DS' in lv2_group and 'measured_V_GS' in lv2_group:
                # Extract measured I_DS and calculated V_GS
                IDS_data = np.array(lv2_group['measured_I_DS'])
                VGS_data = np.array(lv2_group['measured_V_GS'])

                # Ensure that both arrays are of the same length
                if len(IDS_data) == len(VGS_data):
                    # Stack the two arrays into a two-column format
                    combined_data = np.column_stack((VGS_data, IDS_data))

                    # Define the output filename
                    output_filename = f'{lv1_group_name}__{lv2_group_name}__VGS_IDS.csv'

                    # Save the combined data as a CSV file
                    np.savetxt(output_filename, combined_data, delimiter=',', header="measured VGS, measured IDS", comments='')

                else:
                    print(f"Warning: Length mismatch between VGS and IDS for {lv1_group_name}/{lv2_group_name}")
```

```python
# Extract transfer and output CSV files from HDF5
# Transfer: VGS vs IDS
# Output: VDS vs IDS

import h5py
import numpy as np

# Specify the name of your HDF5 file
filename = "1_2025-12-15 Day1 Mixing -C6,C8,C10 with&without PS.hdf5"

# Opening and reading the HDF5 file
with h5py.File(filename, 'r') as file:
    for lv1_group_name in file.keys():  # Loop over level 1 groups
        lv1_group = file[lv1_group_name]
        
        for lv2_group_name in lv1_group.keys():  # Loop over level 2 groups
            lv2_group = lv1_group[lv2_group_name]
            
            # Check if IDS data exists (required for both transfer and output)
            if 'measured_I_DS' in lv2_group:
                IDS_data = np.array(lv2_group['measured_I_DS'])
                
                # Extract TRANSFER curves (VGS vs IDS)
                if 'measured_V_GS' in lv2_group:
                    VGS_data = np.array(lv2_group['measured_V_GS'])
                    
                    if len(IDS_data) == len(VGS_data):
                        # Stack the two arrays into a two-column format
                        transfer_data = np.column_stack((VGS_data, IDS_data))
                        # Define the output filename
                        transfer_filename = f'{lv1_group_name}__{lv2_group_name}__VGS_IDS.csv'
                        # Save the transfer data as a CSV file
                        np.savetxt(transfer_filename, transfer_data, delimiter=',', 
                                   header="measured_VGS,measured_IDS", comments='')
                        print(f"✓ Transfer curve saved: {transfer_filename}")
                    else:
                        print(f"⚠ Warning: Length mismatch between VGS and IDS for {lv1_group_name}/{lv2_group_name}")
                
                # Extract OUTPUT curves (VDS vs IDS)
                if 'measured_V_DS' in lv2_group:
                    VDS_data = np.array(lv2_group['measured_V_DS'])
                    
                    if len(IDS_data) == len(VDS_data):
                        # Stack the two arrays into a two-column format
                        output_data = np.column_stack((VDS_data, IDS_data))
                        # Define the output filename
                        output_filename = f'{lv1_group_name}__{lv2_group_name}__VDS_IDS.csv'
                        # Save the output data as a CSV file
                        np.savetxt(output_filename, output_data, delimiter=',', 
                                   header="measured_VDS,measured_IDS", comments='')
                        print(f"✓ Output curve saved: {output_filename}")
                    else:
                        print(f"⚠ Warning: Length mismatch between VDS and IDS for {lv1_group_name}/{lv2_group_name}")

print("\n✓ Extraction completed!")
```
